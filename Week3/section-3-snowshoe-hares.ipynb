## Part 5

Steps to data wrangling: 
- convert the lower case f/m to the same thing, stripping them of spaces
- get rid of the question marks after f/m
- turn the ? into missing value
- turn the f/m and F/M into female and male

## Part 6

```python
hares['sex_simple'] = hares['sex'].str.lower()
hares['sex_simple'] = hares['sex_simple'].str[:1]
hares['sex_simple'].value_counts(dropna=False)
```

```python
hares.loc[(hares['sex_simple'] == '?') | (hares['sex_simple'] == 'p')] = np.nan
```

```python
hares['sex_simple'] = hares['sex_simple'].replace(to_replace= 'f', value= 'female')
hares['sex_simple'] = hares['sex_simple'].replace(to_replace= 'm', value= 'male')
```
